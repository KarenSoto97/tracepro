```python
from tracepro.app import TP_app
from tracepro.element import Element
from tracepro.source import Source
```

### New file

```python
working_path =r"C:\Software\tracepro\examples\results\polygon_3D"
tp_file_name = "polygon_3D"

tp = TP_app(macro_path=working_path)
tp.new_file(file_name=tp_file_name, tp_path=working_path)
tp.execute(graphic_window=True)
```

### Hexagon

```python
hexagon = Element(name='Hexagon')
macro_text = hexagon.polygon_3D(num_sides=6, radius=5, height=10, material=("", ""), rotation=(0, 90, 0))
tp.add_function(macro_text)
tp.apply_color(object_name=hexagon.name, color=(255, 0, 0))
tp.execute()
```

### Hollow Hexagon

```python
hollow_hexagon = Element(name='Hollow_Hexagon')
macro_text = hollow_hexagon.polygon_3D(num_sides=6, radius=5, height=10, material=("", ""), wall_thickness=3, center=(0, 0, 10), rotation=(0, 90, 0))
tp.add_function(macro_text)
tp.apply_color(object_name=hollow_hexagon.name, color=(0, 255, 0))
tp.execute()
```

### Hollow Pentagon

```python
hollow_pentagon = Element(name='Hollow_Pentagon')
macro_text = hollow_pentagon.polygon_3D(num_sides=5, radius=5, height=10, material=("", ""), wall_thickness=3, center=(0, 0, 20), rotation=(0, 90, 0))
tp.add_function(macro_text)
tp.apply_color(object_name=hollow_pentagon.name, color=(0, 0, 255))
tp.execute()
```

```python
tp.save_file(file_name=tp_file_name, tp_path=working_path)
tp.execute()
```
