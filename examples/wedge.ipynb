```python
from tracepro.app import TP_app
from tracepro.element import Element
from tracepro.source import Source
```

### New file

```python
working_path =r"C:\Software\tracepro\examples\results\wedge"
tp_file_name = "wedge_example"

tp = TP_app(macro_path=working_path)
tp.new_file(file_name=tp_file_name, tp_path=working_path)
tp.execute(graphic_window=True)
```

### Wedge

```python
wedge = Element(name="wedge")
macro = wedge.wedge(length=5, depth=5, angle=45, center=(0, 0, 0))
tp.add_function(macro)
tp.apply_color(object_name=wedge.name, color=(255, 0, 255), transparency=True)
tp.execute()
```

```python
wedge_2 = Element(name="wedge_2")
macro = wedge_2.wedge(length=5, depth=5, angle=45, center=(0, 0, 0))
tp.add_function(macro)
tp.apply_color(object_name=wedge_2.name, color=(0, 120, 255), transparency=True)
tp.rotate_object(object_name=wedge_2.name, angle=180, axis_type=0, rotation_point=(0, 0, 0), object_ref=True, degrees=True)
tp.move_object(object_name=wedge_2.name, mode="relative", distance=(0, 1, 1))
tp.execute()
```

```python
tp.save_file(file_name=tp_file_name, tp_path=working_path)
tp.execute()
```
